```python
import pandas as pd
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv(r"C:\Users\sriva\Downloads\student_marks_dataset.csv")
```

```python
l=[i for i in range(30)]
```

```python
data['sno']=l
```

```python
data.head(5)
```

```python
data.columns
```

```python
plt.scatter(data['sno'],data['Hours_Studied'])
plt.scatter(data['sno'],data['Assignments'])
plt.scatter(data['sno'],data['Attendance'])
plt.scatter(data['sno'],data['Marks'])
plt.show
```

```python
x=data[['Hours_Studied', 'Assignments', 'Attendance', 'Marks']]
```

```python
x
```

```python
scalling=StandardScaler().fit_transform(x)
```

```python
scalling
```

```python
model=PCA(n_components=2)
```

```python
model
```

```python
result=model.fit_transform(scalling)
```

```python
result
```

```python
plt.scatter(l,result[0: ,0])
plt.scatter(l,result[0: ,1])
plt.show()
```
