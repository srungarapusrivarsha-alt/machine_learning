```python
import pandas as pd
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv(r"C:\Users\sriva\Downloads\Mall_Customers.csv")
```

```python
data
```

```python
data.columns
```

```python
x=data[['Age', 'Annual Income (k$)','Spending Score (1-100)']]
```

```python
x
```

```python
model=DBSCAN(min_samples=4,eps=15)
```

```python
model.fit_predict(x)
```

```python
x['group']=model.fit_predict(x)
```

```python
x
```

```python
plt.scatter(x['Age'],x['Annual Income (k$)'],x['Spending Score (1-100)'],c=x['group'])
plt.show()
```

```python
x=x[x['group']!=-1]
```
