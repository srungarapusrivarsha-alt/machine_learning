```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
data=pd.read_csv(r"C:\Users\sriva\Downloads\Mall_Customers.csv")
```

```python
data.columns
```

```python
x=data[['Age', 'Annual Income (k$)','Spending Score (1-100)']]
```

```python
model=KMeans(n_clusters=4)
```

```python
model.fit(x)
```

```python
model.labels_
```

```python
output=model.predict([[20,38,56]])
```

```python
print(output)
```

```python
group=model.predict(x)
group
```

```python
x['group']=group
```

```python
center=model.cluster_centers_
center
```

```python
center[0::,0:1]
```

```python
center[0::,1::2]
```

```python
center[0::,2::3]
```

```python
plt.scatter(x['Age'],x['Annual Income (k$)'],x['Spending Score (1-100)'],c=x['group'])
plt.scatter(center[0::,0:1],center[0::,1::2],center[0::,2::3],c="r")
plt.show
```
