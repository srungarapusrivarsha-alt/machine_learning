```python
import pandas as pd
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram,linkage
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv(r"C:\Users\sriva\Downloads\Mall_Customers.csv")
```

```python
data
```

```python
x=data[['Age','Annual Income (k$)','Spending Score (1-100)']]
```

```python
x
```

```python
values=linkage(x,method='ward')
```

```python
dendrogram(values)
plt.show()
```

```python
model=AgglomerativeClustering(n_clusters=3,linkage='ward')
```

```python
model
```

```python
model.fit_predict(x)
```

```python
x['group']=model.fit_predict(x)
```

```python
x
```

```python
plt.scatter(x['Age'],x['Annual Income (k$)'],x['Spending Score (1-100)'],c=x['group'])
plt.show()
```
