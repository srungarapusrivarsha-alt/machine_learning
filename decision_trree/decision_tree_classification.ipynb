```python
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.metrics import classification_report
from sklearn.preprocessing import LabelEncoder
```

```python
data=pd.read_excel(r"C:\Users\sriva\Downloads\Buy_Computer_Decision_Tree_Classification_Clean_50_Rows.xlsx")
```

```python
# data["Age"]=LabelEncoder().fit_transform(data["Age"])
# data["Age"]
```

```python
x=data[['Age', 'Income', 'Student', 'Credit_Rating']]
```

```python
y=data['Buy_Computer']
```

```python
x.columns
```

```python
for i in x.columns:
    print(i,"------------->",data[i].unique())
```

```python
x["Age"]=x["Age"].map({"Youth":0,"Middle-aged":1,"Senior":2})
x["Income"]=x["Income"].map({'High':2, 'Medium':1, 'Low':0})
x["Student"]=x["Student"].map({'No':0, 'Yes':1})
x["Credit_Rating"]=x["Credit_Rating"].map({'Fair':1, 'Excellent':2})
```

```python
# for i in x.columns:
#     x[i]=LabelEncoder().fit_transform(x[i])
    
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y)
```

```python
model=DecisionTreeClassifier(criterion="entropy",max_depth=3)
```

```python
model
```

```python
model.fit(x_train,y_train)
```

```python
output=model.predict([[0,0,0,1]])
output
```

```python
train_output=model.predict(x_train)
```

```python
train_report=classification_report(y_train,train_output)
print(train_report)
```

```python
import matplotlib.pyplot as plt
plot_tree(
        model,
        feature_names=x.columns    
)
plt.show()
```
