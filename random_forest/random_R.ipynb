```python
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
import matplotlib.pyplot as plt
import numpy as np
from sklearn.tree import plot_tree 
```

```python
data=pd.read_excel(r"C:\Users\sriva\Downloads\RANDOM_FOREST_R.xlsx")
```

```python
data
```

```python
data['Rainfall_mm']=data["Rainfall_mm"]/100
data["Temperature_C"]=data["Temperature_C"]/10
data["Fertilizer_kg"]=data["Fertilizer_kg"]/10
```

```python
data["Yield_kg_per_acre"]=data["Yield_kg_per_acre"]/1000
```

```python
data.columns
```

```python
x=data[['Rainfall_mm', 'Temperature_C', 'Fertilizer_kg','Soil_Quality', 'Irrigation_Level']]
```

```python
y=data["Yield_kg_per_acre"]
```

```python
x
```

```python
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20)
```

```python
model=RandomForestRegressor(n_estimators=5)
```

```python
model
```

```python
model.fit(x_train,y_train)
```

```python
x.columns
```

```python
output=model.predict([[8.9,7.6,8.5,10,9]])
```

```python
print(output)
```

```python
train_output=model.predict(x_train)
```

```python
print(train_output)
```

```python
train_mae=mean_absolute_error(y_train,train_output)
train_mse=mean_squared_error(y_train,train_output)
train_rmse=np.sqrt(train_mse)
train_r2=r2_score(y_train,train_output)
```
